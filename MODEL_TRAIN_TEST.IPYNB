```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler


```

```python
df = pd.read_csv('youtube_shorts_tiktok_trends_2025.csv_ML.csv')


```

```python
features = ['title_len', 'text_richness', 'like_rate', 'comment_rate', 'share_rate', 'views_per_day']
X = df[features]


```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X)

```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df, 
    x='like_rate', 
    y='views_per_day', 
    hue='cluster', 
    palette='viridis', 
    alpha=0.6
)
plt.title('Video Clusters Based on Engagement (Unsupervised Learning)', fontsize=14)
plt.xlabel('Like Rate', fontsize=12)
plt.ylabel('Views Per Day', fontsize=12)
plt.show()
```
